```python

from kanren import Relation, facts, var, run

# 1. Define relations
is_fruit = Relation()
has_color = Relation()

# 2. Add facts to the knowledge base
facts(is_fruit, "Apple", "Banana", "Cherry", "Mango")

facts(has_color,
      ("Apple", "Red"),
      ("Banana", "Yellow"),
      ("Cherry", "Red"),
      ("Mango", "Yellow"))

# 3. Create a query variable
x = var()

# Query 1: Find all red fruits
red_fruits = run(0, x, is_fruit(x), has_color(x, "Red"))
print("Red fruits:", red_fruits)

# Query 2: Find the color of a Banana
color = var()
banana_color = run(1, color, has_color("Banana", color))
print("Banana color:", banana_color[0])
```
